```python
import numpy as np
import pandas as pd
```

Création de la variable Capital

```python
bdd = pd.read_excel('Base_ER_FR.xlsx')

# CORRECTION : Tx en % → /100
mask = (bdd['Tx'] > 0) & (bdd['NB_ECH'] > 0)
bdd['Px_contrat'] = np.where(
    mask,
    bdd['MTECH'] * (1 - (1 + (bdd['Tx']/100)/12)**(-bdd['NB_ECH'])) / ((bdd['Tx']/100)/12),
    0
)

bdd['interets_totaux'] = bdd['MTECH'] * bdd['NB_ECH'] - bdd['Px_contrat']

# Vérif ligne 0
print("Ligne 0 corrigée :")
print(bdd[['NB_ECH', 'MTECH', 'Tx', 'Px_contrat', 'interets_totaux']].iloc[0])
print("\nStats corrigées :")
print(bdd['Px_contrat'].describe())
```

```python
# ============================================================
# 1. Variable cible IFRS9 : Early Repayment observé (ER_obs)
# ============================================================

# Nettoyage strict du remboursement anticipé
bdd["RA_CLEAN"] = bdd["RA"].clip(lower=0)

# Nettoyage strict de l'exposition
bdd["E_EAD_CLEAN"] = bdd["E_EAD"].clip(lower=0)

# ER marginal observé (définition CAPFM / IFRS9)
bdd["ER_obs"] = np.where(
    (bdd["E_EAD_CLEAN"] > 0) & (bdd["RA_CLEAN"] > 0),
    bdd["RA_CLEAN"] / bdd["E_EAD_CLEAN"],
    0.0
)

# Sécurité : ER ne doit pas exploser numériquement
bdd["ER_obs"] = bdd["ER_obs"].clip(lower=0)

# Indicateur événement (utile mais NON central IFRS9)
bdd["PREPAY_EVENT"] = (bdd["ER_obs"] > 0).astype(int)


# ============================================================
# 2. Cycle de vie du crédit (maturité / horizon)
# ============================================================

bdd["B_MAT"] = bdd["B_MAT"].astype(float)
bdd["B_RESMAT"] = bdd["B_RESMAT"].astype(float)

# Âge du crédit
bdd["AGE_CREDIT"] = bdd["B_MAT"] - bdd["B_RESMAT"]

# Ratio de maturité restante
bdd["MAT_RATIO"] = bdd["B_RESMAT"] / bdd["B_MAT"]

# Sécurités IFRS9
bdd.loc[(bdd["MAT_RATIO"] < 0) | (bdd["MAT_RATIO"] > 1), "MAT_RATIO"] = np.nan
bdd.loc[bdd["AGE_CREDIT"] < 0, "AGE_CREDIT"] = np.nan


# ============================================================
# 3. Exposition (EAD)
# ============================================================

bdd["LOG_EAD"] = np.where(
    bdd["E_EAD_CLEAN"] > 0,
    np.log1p(bdd["E_EAD_CLEAN"]),
    np.nan
)


# ============================================================
# 4. Revenus
# ============================================================

bdd["LOG_REV"] = np.where(
    bdd["MREVTOT"] > 0,
    np.log1p(bdd["MREVTOT"]),
    np.nan
)


# ============================================================
# 5. Comportement de paiement (risque concurrent)
# ============================================================

bdd["IMP_RATIO"] = np.where(
    bdd["NB_ECH"] > 0,
    bdd["NBIMP"] / bdd["NB_ECH"],
    0.0
)

bdd["IMP_RATIO"] = (
    bdd["IMP_RATIO"]
    .replace([np.inf, -np.inf], np.nan)
    .fillna(0)
)


# ============================================================
# 6. Effort de remboursement (DSTI proxy)
# ============================================================

bdd["PAYMENT_INCOME_RATIO"] = np.where(
    bdd["MREVTOT"] > 0,
    bdd["MTECH"] / bdd["MREVTOT"],
    np.nan
)

bdd["PAYMENT_INCOME_RATIO"] = (
    bdd["PAYMENT_INCOME_RATIO"]
    .replace([np.inf, -np.inf], np.nan)
    .fillna(bdd["PAYMENT_INCOME_RATIO"].median())
)


# ============================================================
# 7. Taux contractuel (nettoyage IFRS9)
# ============================================================

bdd["Tx"] = (
    bdd["Tx"]
    .clip(lower=0)
    .clip(upper=bdd["Tx"].quantile(0.995))  # évite écrasement arbitraire à 0.5
)
```

Création variable capital

```python
# CORRECTION : Tx en % → /100
mask = (bdd['Tx'] > 0) & (bdd['NB_ECH'] > 0)
bdd['Capital'] = np.where(
    mask,
    bdd['MTECH'] * (1 - (1 + (bdd['Tx']/100)/12)**(-bdd['NB_ECH'])) / ((bdd['Tx']/100)/12),
    0
)

bdd['interets_totaux'] = bdd['MTECH'] * bdd['NB_ECH'] - bdd['Capital']

# Vérif ligne 0
print("Ligne 0 corrigée :")
print(bdd[['NB_ECH', 'MTECH', 'Tx', 'Capital', 'interets_totaux']].iloc[0])
print("\nStats corrigées :")
print(bdd['Capital'].describe())
```

```python
bdd["ER_obs"]= bdd["RA"] / bdd["Capital"]
```

```python
bdd.loc[bdd["Capital"] <= 0, "ER_obs"] = np.nan
bdd.loc[(bdd["ER_obs"] < 0) | (bdd["ER_obs"] > 1), "ER_obs"] = np.nan
```

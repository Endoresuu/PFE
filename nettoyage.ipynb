```python
import pandas as pd
import numpy as np

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 10)

bdd = pd.read_excel("Base_ER_FR.xlsx")

bdd.head()
```

```python
bdd.info()
```

```python
bdd.describe(include="all")
```

```python
bdd["CLASSACT"] = bdd["CLASSACT"].astype("category")

bdd["CLASSACT"].cat.categories
```

```python
bdd["CLASSACT"].value_counts()
```

```python
bdd["CSP"].value_counts(dropna=False)
bdd["CSP"] = bdd["CSP"].astype("category")
bdd["CSP"].cat.categories
```

```python
bdd["DARRET"] = pd.to_datetime(
    bdd["DARRET"].astype(str) + "01",
    format="%Y%m%d",
    errors="coerce"
)

bdd["DARRET"].head(10)
bdd["DARRET"].describe()
```

```python
bdd["E_OFF"].describe()
```

```python
bdd.drop(columns=["E_OFF"], inplace=True)
bdd.columns
```

```python
np.allclose(
    bdd["E_EAD"],
    bdd["E_ONB"],
    equal_nan=True
)
bdd.drop(columns=["E_ONB"], inplace=True)
bdd.columns
```

```python
(bdd["MREVTOT"] - (bdd["MREVNU"] + bdd["MREVAU"])).describe()
bdd.drop(columns=["MREVNU", "MREVAU"], inplace=True)
bdd.columns
```

```python
bdd["default"] = (bdd["NBIMP"] > 0).astype(int)
bdd["default"].value_counts()
bdd["default"].value_counts(normalize=True)
```

```python
bdd["produit"].value_counts(dropna=False)
bdd["produit"] = bdd["produit"].astype("category")

cats = list(bdd["produit"].cat.categories)
if "AMO" in cats:
    cats.insert(0, cats.pop(cats.index("AMO")))
    bdd["produit"] = bdd["produit"].cat.reorder_categories(cats)

bdd["produit"].cat.categories
```
